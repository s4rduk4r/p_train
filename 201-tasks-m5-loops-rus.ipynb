## Задача 79

Добавьте оператор `break`, где это необходимо, чтобы получить следующий результат:
```
1
1
100
20
10
```

```python
x = [1, 2]
y = [10, 100]
for i in x:
    for j in y:
        if i % 2 == 0:
            print(i * j)
        print(i)
    print(j)
```
