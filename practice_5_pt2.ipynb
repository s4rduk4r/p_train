- **Интегралы**

$$ \int_{a}^{b} f(x) dx $$

Код LaTex:
```latex
$$ \int_{a}^{b} f(x) dx $$
```

Это определение интеграла. Здесь $a$ и $b$ - нижняя и верхняя граница интегрирования соответственно, а $f(x)$ - функция, которую нужно проинтегрировать.

- **Матрицы**

$$
A = \begin{pmatrix}
a & b \\
c & d
\end{pmatrix}
$$

Код LaTex:
```latex
$$
A = \begin{pmatrix}
a & b \\
c & d
\end{pmatrix}
$$
```

Это матрица размером 2x2. Элементы матрицы обозначаются как $a$, $b$, $c$ и $d$.

- **Символьные операции**

$$
\sum_{i=1}^n i = \frac{n(n+1)}{2}
$$

Код LaTex:
```latex
$$
\sum_{i=1}^n i = \frac{n(n+1)}{2}
$$
```

Это формула суммы арифметической прогрессии. Она показывает, как вычислить сумму первых $n$ целых чисел.

Используя справку о математических выражениях запишите формулы для:
- индекса массы тела (ИМТ) 
- расчёта стоимости пребывания в стационаре по известной цене койко-дня, количеству дней и количеству пациентов.
